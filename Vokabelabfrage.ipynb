```python
german = "Raumschiff"
english = "spaceship"
```

```python
answer = input(german)

if answer == english:
    print("correct")
else:
    print("wrong")
    print("Your answer was: " + answer)
    print("The correct answer is: " + english)
```

```python
vocabulary = {
    "Raumschiff" : "spaceship",
    "Planet" : "planet",
    "Astronaut" : "astronaut",
    "Stiefel" : "boots",
    "staub" : "dust",
    "Sonne" : "sun",
    "Mond" : "moon",
    "Zug" : "train",
}
```

```python
for german, english in vocabulary.items():
    answer = input(german)

    if answer == english:
        print("👍 correct")

    else:
        print("😥 wrong")
        print("Your answer was: " + answer)
        print("The correct answer is: " + english)
```
